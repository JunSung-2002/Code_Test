# while 문

```python
treeHit = 0 #<- 나무를 찍은 횟수
while treeHit < 10: #<- 조건
    treeHit += 1
    print("나무를 %d 번 찍었습니다."% treeHit)
    if treeHit == 10:
        print("나무 넘어 갑니다.")
```

```python
anser = 0
while anser < 9:
  anser += 1
  print("말이 너무 많아요. %d 번 말하고 있어요"% anser)
  if anser == 9:
    print('나 이제 바빠요')
```

**break 사용하기**

```python
coffee = 10
money = 300
while money:
    print("돈을 받았으니 커피를 줍니다.")
    coffee = coffee  -1
    print("남은 커피양은 %d개입니다." % coffee)
    if coffee == 0:
        print("커피가 다 떨어졌습니다. 판매를 중지합니다")
        break
```

```python
coke = 10
point = 200
while point:
  print("포인트 10이 차감되었습니다.")
  coke = coke -1
  print("남은 콜라는 %d개입니다." % coke)
  if coke == 1:
    print("콜라가 다 떨어졌습니다. 판매를 중지합니다")
    break
```

**continue 사용하기**

```python
a= 0
while a <100:
    a = a+3
    if a % 2 ==0: #<- % = 나머지, 나머지가 0이면 다시 a+1로 돌아감
        continue
    print(a)
```

**리스트를 이용한 for 반복문**

```python
buy_list = ['coke', 'sugar', 'candy']
for o in buy_list:
  print(o)
  if 'sugar' in o:
    print("***")
```

**내포 리스트**

```python
a = [1, 2, 3, 4]
result = [num * 3 for num in a if num % 2 == 0]
print(result)
```

```python
a = [1,2,3,4]
result = [num * 3 for num in a]
print(result)
```

**구구단의 모든 결과를 리스트로 담는 리스트 내포 예제**

```python
result = [x*y for x in range(2,19) for y in range(1,10)]
print(result)
```

# 학생들의 성적 처리 하기

```python
score = [[23,90,20,90],[25,50,20,10],[15,60,5,70],[0,30,20,20],[0,10,0,30],[5,70,5,100]]

grade = []

for i in score: #학생 개인의 성적

    sum = 0
    for j in i: #성적 반영의 요소 (퀴즈, 중간고사, 레포트, 기말 고사)
        sum +=j
    grade.append(sum)

print(grade)
```

```python
point = [[100,90,80,100,100],[100,80,85,80,100],[90,90,90,100,80],[100,70,60,40,50],[100,100,100,20,100]]

level = []

for p in point:

    sum = 0
    for j in p:
        sum +=j
    level.append(sum)
print(level)
```
